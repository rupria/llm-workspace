```python
from langchain_community.retrievers import BM25Retriever
from langchain_core.documents import Document

# 1. 문서 샘플 준비
docs = [
    Document(page_content="금융투자협회 모바일 앱 로그인 오류 해결 안내"),
    Document(page_content="주식 위탁 수수료 이벤트 안내"),
    Document(page_content="FAISS 벡터 DB 구축 가이드 문서")
]

# 2. BM25 Retriever 생성 및 형태소/단어 분할 인덱싱
retriever = BM25Retriever.from_documents(docs)
retriever.k = 2  # 가져올 문서 개수 설정

# 3. 키워드 검색 수행
results = retriever.invoke("모바일 앱")
```
